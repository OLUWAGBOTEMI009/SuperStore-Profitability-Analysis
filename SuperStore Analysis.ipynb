## Exploratory Data Analysis

**Q1. Which category generates the most profit?**

```python
# Profit by category with margin
cat_profit = df.groupby("Category")[["Sales", "Profit"]].sum().sort_values("Profit")


# Plot
plt.figure(figsize=(8, 4))
plt.barh(cat_profit.index, cat_profit["Profit"], color='green')


plt.title("Profit by Category")
plt.xlabel("Total Profit (USD)")
plt.tight_layout()
plt.savefig("profit_by_category_simple.png", dpi=150)
plt.show()
```

**Technology generates the highest profit.**

**Q2. Which region has high sales but low profit margin?**

```python
region = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .reset_index()
    .sort_values("Sales", ascending=False)
)
region["Margin %"] = region["Profit"] / region["Sales"] * 100

# Create chart
fig, ax1 = plt.subplots(figsize=(9, 4.5))
ax2 = ax1.twinx()

x = range(len(region))
ax1.bar(x, region["Sales"], color="green", label="Sales")
ax2.plot(x, region["Margin %"], color="orange", marker="o", linewidth=2, markersize=6, label="Profit margin %")

ax1.set_xticks(x)
ax1.set_xticklabels(region["Region"])
ax1.set_ylabel("Total Sales")
ax2.set_ylabel("Profit Margin (%)")



# Combine legends
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right", frameon=False)

plt.title("Sales vs Profit Margin by Region")
plt.tight_layout()
plt.savefig("region_sales_vs_margin.png", dpi=150, bbox_inches="tight")
```

```python
# Identify the region with highest sales
top_sales_region = region.iloc[0]["Region"]
lowest_margin_region = region.loc[region["Margin %"].idxmin(), "Region"]

print("Region summary:")
print(region)
print(f"\n Region with highest sales: {top_sales_region}")
print(f" Region with lowest profit margin: {lowest_margin_region}")
```

**The Central region has the lowest profit margin while West has the highest sales.**

**Q3. Does high discount always lead to negative profit?**

```python
# Group by discount level (rounded to 5% intervals)
df["discount_bin"] = (df["Discount"] // 0.05) * 0.05
discount_effect = df.groupby("discount_bin").agg(
    avg_profit=("Profit", "mean"),
    count=("Profit", "count")
).reset_index()

# Show where profit becomes negative
print("Average profit by discount level:")
print(discount_effect)
negative_at = discount_effect[discount_effect["avg_profit"] < 0].iloc[0]["discount_bin"]
print(f"\n Average profit turns negative when discount ≥ {negative_at:.0%}")

# Chart
plt.figure(figsize=(8,4))
plt.bar(discount_effect["discount_bin"], discount_effect["avg_profit"], width=0.04, color="tomato", alpha=0.7)
plt.axhline(y=0, color="black", linestyle="--", linewidth=0.8)
plt.xlabel("Discount (%)")
plt.ylabel("Average Profit per Order ($)")
plt.title("Does High Discount Kill Profit?")
plt.tight_layout()
plt.savefig("discount_vs_profit.png", dpi=150)
plt.show()
```

**Typically when the discount is >= 25%, average profit becomes negative**

**Q4. Which product subcategories are losing money?**

```python
# Group by Sub-Category
subcat_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values()

# Show top 5 losers
losers = subcat_profit[subcat_profit < 0].head(5)
print("Product sub‑categories with negative total profit:")
print(losers)

# Chart
plt.figure(figsize=(10,5))
colors = ["#D9534F" if v < 0 else "#5CB85C" for v in subcat_profit.values]
subcat_profit.plot(kind="barh", color=colors)
plt.axvline(x=0, color="black", linewidth=0.8)
plt.xlabel("Total Profit ($)")
plt.title("Profit by Product Sub‑Category")
plt.tight_layout()
plt.savefig("profit_by_subcategory.png", dpi=150)
plt.show()
```

**Sub-Categories products like tables, bookcases, and Supplies are losing money**

**Q5. What does the monthly sales trend look like?**

```python
# Convert Order Date to datetime and extract year‑month
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["YearMonth"] = df["Order Date"].dt.to_period("M")

monthly_sales = df.groupby("YearMonth")["Sales"].sum().reset_index()
monthly_sales["YearMonth_str"] = monthly_sales["YearMonth"].astype(str)

# Answer: show peak month
peak_month = monthly_sales.loc[monthly_sales["Sales"].idxmax(), "YearMonth_str"]
print(f" Highest sales month: {peak_month} (${monthly_sales['Sales'].max():,.0f})")

# Chart
plt.figure(figsize=(12,4))
plt.plot(monthly_sales["YearMonth_str"], monthly_sales["Sales"], marker="o", linestyle="-", color="teal")
plt.xticks(rotation=45, ha="right", fontsize=8)
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.title("Monthly Sales Trend (Seasonality Check)")
plt.tight_layout()
plt.savefig("monthly_sales_trend.png", dpi=150)
plt.show()
```

**November–December peak (holiday season), lowest in January–February**

# Key Findings 

1. **Technology generates the highest profit** among all product categories, while Furniture and Office Supplies show lower margins.

2. **Discounts above 25% lead to negative average profit** per order, with high discounts (>40%) consistently producing losses.

3. **Tables, Bookcases, and Supplies** are the only sub‑categories with negative total profit, making them candidates for pricing or sourcing review.
